```python
import pandas as pd
import re
```

```python
# defining input and output files 
input="credit_txn_v5.xlsx"
output="output_file.xlsx"
```

```python
df=pd.read_excel(input)
```

```python
# --- helper to clean illegal Excel characters ---
def clean_excel_text(val):
    if isinstance(val, str):
        return re.sub(r'[\x00-\x08\x0B-\x0C\x0E-\x1F]', '', val)
    return val

# --- compute company frequency per ledger ---
ledger_company_freq = (
    df
    .assign(**{
        'Ledger Name': df['Ledger Name'].map(clean_excel_text)
    })
    .groupby('Ledger Name')['Company ID']
    .nunique()
    .reset_index(name='company_freq')
    .sort_values('company_freq', ascending=False)
)

# --- export to Excel ---
ledger_company_freq.to_excel(
    output,
    index=False
)
```
