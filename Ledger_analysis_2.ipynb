```python
# Import required libraries
import pandas as pd  # Data manipulation and Excel I/O
import re             # Regular expressions for text cleaning
```

```python
# Define input and output file paths
# Input: Excel file containing credit transaction data
# Output: Excel file with ledger-to-company frequency analysis
input = "credit_txn_v5.xlsx"
output = "output_file.xlsx"
```

```python
# Load transaction data from Excel file
df = pd.read_excel(input)
```

```python
# Helper function to clean illegal Excel characters from text
# Excel does not allow control characters (0x00-0x08, 0x0B-0x0C, 0x0E-0x1F) in cell values
def clean_excel_text(val):
    if isinstance(val, str):
        # Remove all control characters that are illegal in Excel
        return re.sub(r'[\x00-\x08\x0B-\x0C\x0E-\x1F]', '', val)
    return val

# Calculate the frequency of unique companies per ledger name
# This shows how many different companies use each ledger account
# Higher frequency indicates a commonly shared ledger across companies

ledger_company_freq = (
    df
    .assign(**{
        # Clean ledger names of illegal Excel characters before processing
        'Ledger Name': df['Ledger Name'].map(clean_excel_text)
    })
    # Group by Ledger Name and count the number of unique Company IDs
    .groupby('Ledger Name')['Company ID']
    .nunique()                                                     # Count distinct companies for each ledger
    .reset_index(name='company_freq')                              # Convert to DataFrame with company frequency column
    .sort_values('company_freq', ascending=False)                  # Sort by frequency in descending order
)

# Export the ledger-company frequency analysis to Excel
ledger_company_freq.to_excel(
    output,
    index=False                                                    # Don't include row indices
)
```
